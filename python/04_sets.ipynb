[**Introduction to Sets**](https://www.hackerrank.com/challenges/py-introduction-to-sets/problem)

```python
def average(array):
    array = set(arr)
    return sum(array)/len(array)

if __name__ == '__main__':
    n = int(input())
    arr = list(map(int, input().split()))
    result = average(arr)
    print(result)
```

[**No Idea!**](https://www.hackerrank.com/challenges/no-idea/problem)

```python
n, m = map(int, input().split())

array = list(input().split())
a = set(input().split())
b = set(input().split())

score = 0

for i in array:
    if i in a:
        score += 1
    elif i in b:
        score -= 1

print(score)
```

[**Symmetric Difference**](https://www.hackerrank.com/challenges/symmetric-difference/problem)

```python
m = input()
a = set(map(int, input().split()))
n = input() 
b = set(map(int, input().split()))

union = a | b
intersection = a & b

symmetric_diff = sorted(list(union - intersection))

for i in symmetric_diff:
    print(i)
```

[**Set .discard(), .remove() & .pop()**](https://www.hackerrank.com/challenges/py-set-discard-remove-pop/problem)

```python
n = int(input())
s = set(map(int, input().split()))
commands_no = int(input())

commands = {
    "pop": s.pop,
    "remove": s.remove,
    "discard": s.discard,
}

for _ in range(commands_no):
    command, *arguments = input().split()
    arguments = tuple(map(int, arguments))

    commands[command](*arguments)

print(sum(s))
```

[**The Captain's Room**](https://www.hackerrank.com/challenges/py-the-captains-room/problem)

```python
k = int(input())
rooms = list(map(int, input().split()))

unique_rooms = set(rooms)

for room in unique_rooms:
    rooms.remove(room)

captain_room = (set(unique_rooms) - set(rooms)).pop()

print(captain_room)
```

[**Check Subset**](https://www.hackerrank.com/challenges/py-check-subset/problem)

```python
test_no = int(input())

for _ in range(test_no):
    n_a = input()
    set_a = set(map(int, input().split()))
    n_b = input()
    set_b = set(map(int, input().split()))

    if len(set_a - set_b):
        print('False')
    else:
        print('True')
```

[**Check Strict Superset**](https://www.hackerrank.com/challenges/py-check-strict-superset/problem)

```python
superset = set(map(int, input().split()))

sets_n = int(input())

is_superset = True

for _ in range(sets_n):
    secondary_set = set(map(int, input().split()))

    if not (secondary_set < superset):
        is_superset = False
        break

print(is_superset)
```
