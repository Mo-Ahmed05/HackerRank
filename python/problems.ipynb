[**Nested Lists**](https://www.hackerrank.com/challenges/nested-list/problem)

````python
def first_solution(students):
    scores = sorted([score for name, score in students])

    for score in scores:
        if score > scores[0]:
            second_lowest_score = score
            break

    result_names = [
        name for name, score in students 
        if score == second_lowest_score]
    result_names.sort()

    return result_names



# Better Soulution
def second_solution(students):
    scores = sorted(set(score for name, score in students))

    second_lowest_score = scores[1]

    result_names = sorted([
        name for name, score in students 
        if score == second_lowest_score])

    return result_names



students = []
for _ in range(int(input())):
    name = input()
    score = float(input())
    students.append([name, score])

result_names = second_solution(students)

for name in result_names:
    print(name)
````

[**sWAP cASE**](https://www.hackerrank.com/challenges/swap-case/problem)

````python
def swap_case(s):

    chars = list(s)
    swapped = ""

    for i in range(len(chars)):

        if chars[i].islower():
            chars[i] = chars[i].upper()
        else:
            chars[i] = chars[i].lower()

        swapped += chars[i]

    return swapped

if __name__ == '__main__':
    s = input()
    result = swap_case(s)
    print(result)
````

[**Find a string**](https://www.hackerrank.com/challenges/find-a-string/problem)

````python
def count_substring(string, sub_string):
    n = 0

    for i in range(len(string)-len(sub_string)+1):
        if string[i:i+len(sub_string)] == sub_string:
            n += 1

    return n



if __name__ == '__main__':
    string = input().strip()
    sub_string = input().strip()
    
    count = count_substring(string, sub_string)
    print(count)
````

[**String Validators**](https://www.hackerrank.com/challenges/string-validators/problem)

````python
s = input()

hasAlphanumeric = False
hasAlphabetical = False
hasDigits = False
hasLowercase = False
hasUppercase = False

for ch in s:
    if not hasAlphanumeric:
        hasAlphanumeric = ch.isalnum()
    if not hasAlphabetical:
        hasAlphabetical = ch.isalpha()
    if not hasDigits:
        hasDigits = ch.isdigit()
    if not hasLowercase:
        hasLowercase = ch.islower()
    if not hasUppercase:
        hasUppercase = ch.isupper()

    if (hasAlphanumeric and
        hasAlphabetical and 
        hasDigits and
        hasLowercase and hasUppercase):
        break

print(hasAlphanumeric)
print(hasAlphabetical)
print(hasDigits)
print(hasLowercase)
print(hasUppercase)
````

[**Text Alignment**](https://www.hackerrank.com/challenges/text-alignment/problem)

````python
thickness = int(input()) #This must be an odd number
c = 'H'

#Top Cone
for i in range(thickness):
    print((c*i).rjust(thickness-1)+c+(c*i).ljust(thickness-1))

#Top Pillars
for i in range(thickness+1):
    print((c*thickness).center(thickness*2)+(c*thickness).center(thickness*6))

#Middle Belt
for i in range((thickness+1)//2):
    print((c*thickness*5).center(thickness*6))    

#Bottom Pillars
for i in range(thickness+1):
    print((c*thickness).center(thickness*2)+(c*thickness).center(thickness*6))    

#Bottom Cone
for i in range(thickness):
    print(((c*(thickness-i-1)).rjust(thickness)+c+(c*(thickness-i-1)).ljust(thickness)).rjust(thickness*6))
````

[**String Formatting**](https://www.hackerrank.com/challenges/python-string-formatting/problem)

````python
def print_formatted(number):
    width = len(bin(number)) - 2

    for i in range(1, number+1):

        print(
            str(i).rjust(width) + " " +
            oct(i)[2:].rjust(width) + " " +
            hex(i)[2:].upper().rjust(width) + " " +
            bin(i)[2:].rjust(width)
            )


if __name__ == '__main__':
    n = int(input())
    print_formatted(n)
````

[**Alphabet Rangoli**](https://www.hackerrank.com/challenges/alphabet-rangoli/problem)

````python
def print_rangoli(size):

    alphabets = [
        'a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm',
        'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z'
        ]

    for i in range(size):
        alpha_line = sorted(alphabets[size-i-1:size], reverse=True)+alphabets[size-i:size]
        print('-'*(size-1-i)*2 + '-'.join(alpha_line) + '-'*(size-1-i)*2)

    for i in range(size-2, -1, -1):
        alpha_line = sorted(alphabets[size-i-1:size], reverse=True)+alphabets[size-i:size]
        print('-'*(size-1-i)*2 + '-'.join(alpha_line) + '-'*(size-1-i)*2)

if __name__ == '__main__':
    n = int(input())
    print_rangoli(n)
````

[**The Minion Game**](https://www.hackerrank.com/challenges/the-minion-game/problem)

````python
def minion_game(string):
    kevin_core = 0
    stuart_core = 0

    vowels = ['A', 'E', 'I', 'O', 'U']

    for i in range(len(string)):

        if string[i] in vowels:
            kevin_score += len(string) - i

        else:
            stuart_Score += len(string) - i

    if kevin_core > stuart_core:
        print(f'Kevin {kevin_core}')
    elif kevin_core < stuart_core:
        print(f'Stuart {stuart_core}')
    else:
        print('Draw')



if __name__ == '__main__':
    s = input()
    minion_game(s)
````

[**Merge the Tools!**](https://www.hackerrank.com/challenges/merge-the-tools/problem)

````python
def merge_the_tools(string, k):

    for i in range(0, len(string), k):
        t = list(string[i:i+k])
        seen = {char: False for char in t}

        for j in range(len(t)):
            if not seen[t[j]]:
                seen[t[j]] = True
            else:
                t[j] = ''

        print(''.join(t))


if __name__ == '__main__':
    string, k = input(), int(input())
    merge_the_tools(string, k)
````
