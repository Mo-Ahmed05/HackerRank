[**List Comprehensions**](https://www.hackerrank.com/challenges/list-comprehensions/problem)

````python
x = int(input())
y = int(input())
z = int(input())
n = int(input())

result = []

for i in range(x+1):
    for j in range(y+1):
        for k in range(z+1):
            if i+j+k != n:
                result.append([i,j,k])

print(result)
````

[**Nested Lists**](https://www.hackerrank.com/challenges/nested-list/problem)

````python
def first_solution(students):
    scores = sorted([score for name, score in students])

    for score in scores:
        if score > scores[0]:
            second_lowest_score = score
            break

    result_names = [
        name for name, score in students 
        if score == second_lowest_score]
    result_names.sort()

    return result_names



# Better Soulution
def second_solution(students):
    scores = sorted(set(score for name, score in students))

    second_lowest_score = scores[1]

    result_names = sorted([
        name for name, score in students 
        if score == second_lowest_score])

    return result_names



students = []
for _ in range(int(input())):
    name = input()
    score = float(input())
    students.append([name, score])

result_names = second_solution(students)

for name in result_names:
    print(name)
````

[**Finding the percentage**](https://www.hackerrank.com/challenges/finding-the-percentage/problem)

````python
n = int(input())
student_marks = {}

for _ in range(n):
    name, *line = input().split()
    scores = list(map(float, line))
    student_marks[name] = scores
query_name = input()

avg = sum(student_marks[query_name]) / len(student_marks[query_name])
print(f'{avg:.2f}')
````

[**Lists**](https://www.hackerrank.com/challenges/python-lists/problem)

````python
N = int(input())
my_list = []

commands = {
    "append": my_list.append,
    "pop": my_list.pop,
    "sort": my_list.sort,
    "reverse": my_list.reverse,
    "remove": my_list.remove,
    "insert": my_list.insert,
    "print": lambda: print(my_list)
}


for _ in range(N):
    command, *arguments = input().split()
    arguments = tuple(map(int, arguments))

    commands[command](*arguments)
````
